### Objective
Write a Python program that evaluates a purchase amount using an `if-else` conditional statement to determine discount eligibility.

### Program Logic
* **Input:** Prompts the user to enter the `total_price` as a floating-point number.
* **Condition:** Checks if `total_price` is greater than or equal to **\$50.00**.
* **Output:**
  * If **True**, calculates a **10% discount** (`total_price * 0.10`) and displays the formatted discount amount to two decimal places.
  * If **False**, informs the user that they do not qualify for a discount.

```python
total_price = float(input("Enter the total price: $"))

if total_price >= 50:
    discount = total_price * 0.10
    final_price = total_price - discount
    print(f"Your discount amount is: ${discount:.2f}")
    print(f"Your final price after discount is: ${final_price:.2f}")
else:
    discount = 0.00
    final_price = total_price - discount
    print("You do not qualify for a discount.")
    print(f"Your final price is: ${final_price:.2f}")
```
