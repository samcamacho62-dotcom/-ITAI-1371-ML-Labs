```python
score = float(input("Enter the exam score: "))

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
elif score >= 60:
    print("D")
else:
    print("F")
```

### Grading System Scale

Here is how the scores map to letter grades based on the code:

* **A**: 90 and above
* **B**: 80 to 89.9
* **C**: 70 to 79.9
* **D**: 60 to 69.9
* **F**: Below 60
